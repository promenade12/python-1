## 热身：定义简单函数
要求：
1. `say_hi`：无参数，打印 `"Hello, world!"`
2. `goat_pad(s)`：接收字符串，前后拼接字符串`goat`后打印
3. `print_n(s, n)`：接收字符串`s`和整数`n`，将字符串打印`n`次，每次单独一行



```python
# 1. say_hi：无参，打印 Hello, world!
def say_hi():
    print("Hello, world!")

# 2. goat_pad：接收字符串，前后拼接 goat
def goat_pad(s):
    print(f"goat{s}goat")

# 3. print_n：字符串s，整数n，打印s共n次，每行一条
def print_n(s, n):
    for _ in range(n):
        print(s)


# --------测试调用--------
say_hi()
print("-"*20)
goat_pad("apple")
print("-"*20)
print_n("test line", 3)
```

## 欧几里得算法
实现`gcd(m,n)`函数求两个整数最大公约数。
1. 使用欧几里得辗转相除法实现gcd函数
2. 测试三组：`(2018, 2019)`、`(1200, 300)`、`(5040, 60)`
3. 测试负数输入，分析行为是否合理。
> 最大公约数：能够同时整除两个数字的最大正整数。


```python
def gcd(m, n):
    # 欧几里得辗转相除法
    while n != 0:
        m, n = n, m % n
    return m


# 测试三组数字对
print("gcd(2018,2019) =", gcd(2018, 2019))
print("gcd(1200,300) =", gcd(1200, 300))
print("gcd(5040,60) =", gcd(5040, 60))

# 测试负数
print("gcd(-24, -18) =", gcd(-24, -18))
```

### 负数输入分析

数学上最大公约数定义针对非负整数，虽然代码可以处理负数，但传入负数不符合GCD数学定义。


## 近似欧拉数 e

$$
e=\sum_{k=0}^{\infty}\frac{1}{k !},\quad e=\lim_{x\to\infty}\left(1+\frac{1}{x}\right)^x
$$

1. `euler_limit(n)`：利用极限公式 $\displaystyle\left(1+\frac{1}{n}\right)^n$ 近似e
2. `euler_infinite_sum(n)`：级数前n项求和。n=0返回0；n=1返回1；`euler_infinite_sum(4)`返回 $1+\frac11+\frac12+\frac16$
3. `euler_approx($\epsilon$)`：while循环累加级数，直到新增项小于误差$\epsilon$，返回e近似值
4. `print_euler_sum_table(n)`：打印k从1~n每一项的级数近似结果，每行输出一组


```python
import math

# 3‑1 euler_limit 极限近似
def euler_limit(n):
    return (1 + 1 / n) ** n

# 3‑2 euler_infinite_sum 级数前n项和
def euler_infinite_sum(n):
    """返回e级数前n项之和，n为非负整数"""
    total = 0.0
    fact = 1  # 0! = 1
    for k in range(n):
        if k != 0:
            fact = fact * k
        total += 1.0 / fact
    return total

# 3‑3 euler_approx 根据误差epsilon近似e
def euler_approx(epsilon):
    total = 0.0
    k = 0
    fact = 1
    while True:
        term = 1.0 / fact
        if term < epsilon:
            break
        total += term
        k += 1
        fact = fact * k
    return total

# 3‑4 print_euler_sum_table 打印表格
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")


# 测试
print(f"euler_limit(10000) = {euler_limit(10000)}")
print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}")
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")
print(f"math.exp(1) 真实e = {math.exp(1)}")
print(f"euler_approx(1e-6) = {euler_approx(1e-6)}")
print("-"*30)
print_euler_sum_table(6)
```

## 测试整数属性
> 正整数n是2的幂：满足 \(n=2^p,\ p\ge0\)。
1. `is_power_of_2(n)`：输入正整数，只用除法与取模`%`，返回布尔值，判断是否为2的幂，禁止使用`math.sqrt`
2. `is_power(b, n)`：泛化函数，参数顺序`(b,n)`，判断n是否是b的某个非负整数次幂。


```python
# 4‑1 判断是否为2的幂
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

# 4‑2 判断 n 是否是 b 的幂
def is_power(b, n):
    if n == 1:
        return True
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

# 测试
print(is_power_of_2(1))   # True 2^0
print(is_power_of_2(8))   # True
print(is_power_of_2(12))  # False
print(is_power_of_2(16))  # True
print("-"*20)
print(is_power(3, 9))   # True 3^2
print(is_power(5, 125)) # True
print(is_power(2, 10))  # False
print(is_power(4,1))    # True 任何数0次方=1
```
