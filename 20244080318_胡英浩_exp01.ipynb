## 实验2：函数、条件、循环与递归

任务：
1. `score_level(score)`函数：
   - 90~100：优秀；80~90：良好；60~80：及格；0~60：不及格
   - 分数不在0‑100抛出`ValueError`
2. 分别使用`for`、`while`计算1~100整数和，对比结果。
3. 迭代版、递归版阶乘函数，约定 $0! = 1$；测试 n=0,1,5,10。
4. `summary(values)`接收非空数值列表，返回元组(最小值,最大值,平均值)，演示元组解包。
5. `sum_all(*args)`可变参数函数，返回所有参数之和；测试无参、单参数、多参数。


```python
# 1. score_level
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

print("score_level测试：")
print(score_level(95), score_level(88), score_level(72), score_level(45))

# 2. for while求1‑100和
sum_for = 0
for i in range(1,101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print(f"\nfor循环1‑100和：{sum_for}")
print(f"while循环1‑100和：{sum_while}")

# 3.阶乘：迭代、递归
def fact_iter(n):
    res = 1
    for k in range(2, n+1):
        res *= k
    return res

def fact_recur(n):
    if n == 0 or n == 1:
        return 1
    return n * fact_recur(n-1)

print("\n阶乘测试：")
for n in [0,1,5,10]:
    print(f"n={n},迭代={fact_iter(n)},递归={fact_recur(n)}")

# 4. summary函数，元组解包
def summary(values):
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

vals = [10,20,30,40]
mi, ma, av = summary(vals)
print(f"\nsummary解包：min={mi},max={ma},avg={av}")

#5. sum_all可变参数
def sum_all(*args):
    return sum(args)

print("\nsum_all测试：")
print("无参", sum_all())
print("单参", sum_all(10))
print("多参", sum_all(1,2,3,4))
```

## 实验4：文件读写、异常处理与模块

任务：
1. 使用`os.getcwd()`查看工作目录；`os.path.join()`构造路径；`os.makedirs(...,exist_ok=True)`创建`exp01_data`文件夹。
2. `with open`写入`students.txt`，再读取，跳过表头，解析文本行。
3. 编写`parse_student(line)`：解析单行；校验字段数、姓名非空、分数0‑100；非法抛出`ValueError`。
4. 逐行处理文件，捕获异常，记录行号错误；保存合法记录为字典列表。
5. 将有效记录数、无效记录数、平均分、及格率写入`exp01_data/report.txt`；演示`w`覆盖模式、`a`追加模式区别。
6. `pickle`序列化保存/读取`students.pkl`，校验数据一致。
7. 将`parse_student`函数保存到外部`student_utils.py`，import导入调用，测试有效、无效输入。
8. 读取不存在文件，捕获`FileNotFoundError`。

> students.txt内容：


```python
import os
import pickle

#1 目录处理
cwd = os.getcwd()
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"当前工作目录：{cwd}")
print(f"数据目录路径：{data_dir}")

txt_path = os.path.join(data_dir, "students.txt")

#2 写入students.txt
content_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content_text)

#3 解析单行学生
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name, score_str = parts
    name = name.strip()
    if not name:
        raise ValueError("姓名为空")
    score = int(score_str.strip())
    if not (0 <= score <= 100):
        raise ValueError(f"分数{score}超出0‑100范围")
    return {"name": name, "score": score}

#4 逐行读取，捕获异常
valid_records = []
error_log = []
with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
for lineno, raw_line in enumerate(lines, start=1):
    if lineno == 1:
        continue #跳过表头
    try:
        rec = parse_student(raw_line)
        valid_records.append(rec)
    except Exception as e:
        error_log.append({"line": lineno, "err": str(e)})

valid_cnt = len(valid_records)
invalid_cnt = len(error_log)
print(f"\n有效记录：{valid_cnt},无效记录：{invalid_cnt}")
for err in error_log:
    print(f"行{err['line']}：{err['err']}")

#统计有效数据
if valid_cnt>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_cnt
    pass_count = sum(1 for r in valid_records if r["score"]>=60)
    pass_rate = pass_count / valid_cnt
else:
    avg_score = 0
    pass_rate = 0

#5 report.txt w模式写入
report_path = os.path.join(data_dir, "report.txt")
report_content = f"""有效记录数：{valid_cnt}
无效记录数：{invalid_cnt}
平均分：{avg_score:.2f}
及格率：{pass_rate:.1%}
"""
with open(report_path, "w", encoding="utf-8") as fw:
    fw.write(report_content)

# a追加模式演示
with open(report_path, "a", encoding="utf-8") as fa:
    fa.write("---追加一行测试内容---\n")
print("\n> w模式覆盖原有全部内容；a模式在文件末尾追加，不删除旧内容")

#6 pickle保存读取
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as fpick:
    pickle.dump(valid_records, fpick)

with open(pkl_path, "rb") as fpick_r:
    loaded_data = pickle.load(fpick_r)
print("\npickle校验：", loaded_data == valid_records)

#7 导出函数到student_utils.py，这里演示写入模块文件
utils_text = '''
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name, score_str = parts
    name = name.strip()
    if not name:
        raise ValueError("姓名为空")
    score = int(score_str.strip())
    if not (0 <= score <= 100):
        raise ValueError(f"分数{score}超出0‑100范围")
    return {"name": name, "score": score}
'''
utils_file = os.path.join(cwd, "student_utils.py")
with open(utils_file, "w", encoding="utf‑8") as fu:
    fu.write(utils_text)

# import外部模块
from student_utils import parse_student
print("\n外部模块导入测试：")
print(parse_student("Tom,78"))
try:
    parse_student("Bad,200")
except ValueError as e:
    print("无效输入异常：", e)

#8 文件不存在捕获FileNotFoundError
try:
    open("not_exist_file.txt","r")
except FileNotFoundError as e:
    print(f"\n捕获文件不存在异常：{e}")
```

## 实验5：类、继承与对象复制

任务：
1. 定义`Student`类：`__init__`初始化`name`、`score`；实现`is_passed()`判断是否及格；实现`__str__()`。
2. 将实验4的有效记录全部实例化为Student对象，输出对象信息与及格状态。
3. `GraduateStudent(Student)`子类，新增`research_topic`属性；`super()`调用父类构造，重写`__str__()`。
4. 对象拷贝：构造带嵌套列表对象 `{"name": "Alice", "scores": [85, 90]}`；分别测试直接赋值、`copy.copy()`浅拷贝、`copy.deepcopy()`深拷贝；修改内部列表，对比原始与副本变化。每次测试从原始对象重新开始。


```python
import copy

#1 Student类定义
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

#2 根据实验4 valid_records生成Student实例
print("Student对象实例：")
stu_objs = []
for rec in valid_records:
    s = Student(rec["name"], rec["score"])
    stu_objs.append(s)
    print(f"{s} 及格：{s.is_passed()}")

#3 GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"

g_stu = GraduateStudent("LiHua", 88, "AI")
print("\n研究生子类：", g_stu, "及格：", g_stu.is_passed())

#4 对象拷贝对比
print("\n====拷贝对比====")
orig = {"name": "Alice", "scores": [85, 90]}

#直接赋值
obj1 = orig
obj1["scores"][0] = 999
print("直接赋值：orig=", orig, "obj1=", obj1)
orig = {"name": "Alice", "scores": [85, 90]} #恢复原始

#浅拷贝 copy.copy
obj2 = copy.copy(orig)
obj2["scores"][0] = 777
print("浅copy.copy：orig=", orig, "obj2=", obj2)
orig = {"name": "Alice", "scores": [85, 90]} #恢复原始

#深拷贝 deepcopy
obj3 = copy.deepcopy(orig)
obj3["scores"][0] = 555
print("深copy.deepcopy：orig=", orig, "obj3=", obj3)
```
